# Ingest dry-run example

Try a load without writing to BigQuery. Install once from `data-eng/`:

```bash
pip install -e "./ingestion[dev]"
```

Set credentials in `data-eng/ingestion/.env` (see `.env.example`).

For reusable runs: `python ingestion/scripts/run_source.py …` or `make ingest-dry`.

````python
import subprocess
import sys

SOURCE = "hdx_wfp_prices"

subprocess.check_call(
    [sys.executable, "-m", "ingestion", "list"],
)
subprocess.check_call(
    [
        sys.executable,
        "-m",
        "ingestion",
        "run",
        SOURCE,
        "--dry-run",
        "--limit",
        "100",
    ],
)
````
